```python
import cv2
import dlib
import matplotlib.pyplot as plt
```

### 1. Завантаження детектора облич

```python
detector = dlib.get_frontal_face_detector()
```

### 2. Функція розпізнавання облич

```python
def detect_faces(img, detector):
    
    faces = detector(img)

    result = img.copy()

    colors = [
        (255, 0, 0),      
        (0, 255, 0),      
        (0, 0, 255),      
        (255, 255, 0),    
        (255, 0, 255), 
        (0, 255, 255)  
    ]

    for i, face in enumerate(faces):
        
        left = face.left()
        top = face.top()
        right = face.right()
        bottom = face.bottom()

        color = colors[i % len(colors)]

        cv2.rectangle(result, (left, top), (right, bottom), color, 3)

    plt.figure(figsize=(10, 7))
    plt.imshow(result)
    plt.title(f'Detected faces: {len(faces)}')
    plt.axis('off')
    plt.show()

    return result
```

### 3. Результат

```python
images = ['face.jpg', 'anime.jpg', 'glasses.jpg', 'faces.jpg', 'street1.jpg', 'street2.jpg']

for filename in images:
    img = cv2.imread(filename)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    result = detect_faces(img, detector)
```
