### Unsharp Masking 

```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
```

### 1. Фільтр підвищення різкості Unsharp Masking
$$
I_{\text{sharp}} =
(amount + 1)*I - amount*I_{\text{blur}}
$$

```python
def unsharp_mask(image, amount):
 
    unsharp = cv2.GaussianBlur(image, (5, 5), 1)
    sharpened = float(amount + 1) * image - float(amount) * unsharp
    sharpened = np.clip(sharpened, 0, 255).astype(np.uint8)
    
    return sharpened
```

### 2. Функція застосування фільтра до зображення

```python
def main(original, amount1, amount2, amount3):
    
    sharpened1 = unsharp_mask(original, amount1)
    sharpened2 = unsharp_mask(original, amount2)
    sharpened3 = unsharp_mask(original, amount3)
    
    plt.figure(figsize=(12, 8))
    
    plt.subplot(221)
    plt.imshow(original)
    plt.title('Original')
    plt.axis('off')
    
    plt.subplot(222)
    plt.imshow(sharpened1)
    plt.title(f'Amount = {amount1}')
    plt.axis('off')
    
    plt.subplot(223)
    plt.imshow(sharpened2)
    plt.title(f'Amount = {amount2}')
    plt.axis('off')
    
    plt.subplot(224)
    plt.imshow(sharpened3)
    plt.title(f'Amount = {amount3}')
    plt.axis('off')
    
    plt.show()
```

### 3. Результат

```python
images = {
    'flower': ('kodim07.png', (1, 2, 4)),
    'tree': ('tree.jpg', (1, 2, 4)),
    'house': ('kodim01.png', (1, 2, 4)),
    'town': ('town.jpg', (2, 4, 6)),
    'bird': ('bird.jpg', (2, 5, 7)),
    'giraf': ('giraf.jpg', (2, 5, 7))
}

for name, (filename, (amount1, amount2, amount3)) in images.items():
    
    original = cv2.imread(filename)
    original = cv2.cvtColor(original, cv2.COLOR_BGR2RGB)

    main(original, amount1, amount2, amount3)
```
