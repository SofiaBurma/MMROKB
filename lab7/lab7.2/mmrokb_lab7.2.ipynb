```python
import os
import cv2
import numpy as np
import pandas as pd
from collections import Counter
from matplotlib import pyplot as plt

plt.rcParams['figure.figsize'] = [15, 10]
```

## **1. Завантаження та підготовка даних**

```python
data = pd.read_csv(os.path.join("dataset", "Train.csv"))
```

```python
num_samples = len(data)

plt.figure(figsize=(15, 8))

for ii in range(15):
    
    idx = np.random.randint(0, num_samples)
    img = cv2.imread(os.path.join(root, data.iloc[idx]["Path"])) 
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    plt.subplot(3, 5, ii + 1)
    plt.imshow(img)
    plt.title(data.iloc[idx]["ClassId"])
    plt.axis("off")

plt.show()
```

## **2. Гістограма кількості зображень за класами**
Для підрахунку використаємо функцію Counter.

```python
hist = Counter(data["ClassId"])

class_counts = pd.Series(hist).sort_index()

display(class_counts.rename("Кількість зображень").to_frame())

plt.figure(figsize=(16, 6))

plt.bar(class_counts.index.astype(str), class_counts.values)

plt.title("Кількість зображень у кожному класі")
plt.xlabel("Номер класу дорожнього знака")
plt.ylabel("Кількість зображень")
plt.xticks(rotation=90)
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()

print("Кількість класів:", len(class_counts))
print("Найбільша кількість зображень у класі:", class_counts.max())
print("Найменша кількість зображень у класі:", class_counts.min())
```
